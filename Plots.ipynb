```python
# Plot NN structure figure
import matplotlib.pyplot as plt

layers = [5, 10, 5, 1]

fig, ax = plt.subplots(figsize=(8, 4))
ax.axis('off')

def get_layer_x(total_layers, i):
    return i * 2

# Draw nodes (as circles, no text)
for i, n_nodes in enumerate(layers):
    x = get_layer_x(len(layers) - 1, i)
    for j in range(n_nodes):
        y = j - (n_nodes - 1) / 2
        circle = plt.Circle((x, y), 0.27, color='mediumseagreen', ec='forestgreen', lw=2, fill=True, zorder=3)
        ax.add_patch(circle)

# Draw lines between layers
for i in range(len(layers) - 1):
    x0 = get_layer_x(len(layers) - 1, i)
    x1 = get_layer_x(len(layers) - 1, i + 1)
    n0, n1 = layers[i], layers[i + 1]
    for j in range(n0):
        for k in range(n1):
            y0 = j - (n0 - 1) / 2
            y1 = k - (n1 - 1) / 2
            plt.plot([x0, x1], [y0, y1], color='mediumseagreen', lw=1.4, zorder=1)

plt.xlim(-1, 7)
plt.ylim(-5, 5)
plt.tight_layout()
plt.savefig('NN_graph', bbox_inches='tight', dpi=150)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

models = [
    "OLS\nRegression",
    "RF",
    "PCA",
    "Neural\nNetwork",
    "Lasso\nRegression",
    "Gradient\nBoosted RF"
]

# Values taken directly from the table:
mae  = [0.75, 0.46, 0.68, 0.95, 1.28, 0.55]
rmse = [1.00, 0.81, 0.92, 1.59, 1.81, 0.96]
r2   = [0.77, 0.85, 0.17, 0.37, 0.18, 0.74]

metrics = ['MAE', 'RMSE', 'R²']
metric_data = [mae, rmse, r2]
colors = ['tab:orange', 'tab:blue', 'tab:green']

x = np.arange(len(models))
width = 0.22

fig, ax = plt.subplots(figsize=(12,7))

for i, (metric, vals) in enumerate(zip(metrics, metric_data)):
    ax.bar(x + i*width - width, vals, width, label=metric, color=colors[i])

ax.set_ylabel('Score')
ax.set_title('Model Performance Comparison')
ax.set_xticks(x)
ax.set_xticklabels(models, fontsize=12)
ax.legend(loc='upper right')
ax.grid(axis='y', alpha=0.5)
plt.tight_layout()
plt.savefig('Model_Comparison.png', bbox_inches='tight', dpi=150)
plt.show()
```
